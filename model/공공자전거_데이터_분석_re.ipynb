```python
import pandas as pd
import numpy as np
```

```python
df_up = pd.read_csv('data/서울특별시 공공자전거 이용정보(월별)_25.1-6.csv')
df_up.head()
```

```python
df_up[['대여소번호', '대여소']] = df_up['대여소명'].str.extract(r'^(\d+)\.\s*(.*)$')
df_up['대여소번호'] = pd.to_numeric(
    df_up['대여소번호'],
    errors='coerce'
)

df_up['대여소'] = df_up['대여소'].str.strip()
df_up.rename(columns={'연령대코드': '연령대'}, inplace=True)

df_up = df_up.drop(['대여소명', '대여구분코드', '성별', '운동량', '탄소량'], axis=1)
```

```python
df_up.head()
```

```python
df_up['대여소번호'] = df_up['대여소번호'].astype('Int64')
df_up.info()
```

```python
df_down = pd.read_csv('data/서울특별시 공공자전거 이용정보(월별)_25.7-12.csv')
df_down.head()
```

```python
df_down[['대여소번호', '대여소']] = df_down['대여소명'].str.extract(r'^(\d+)\.\s*(.*)$')
df_down['대여소번호'] = pd.to_numeric(
    df_down['대여소번호'],
    errors='coerce'
)

df_down['대여소'] = df_down['대여소'].str.strip()
df_down.rename(columns={'연령대코드': '연령대'}, inplace=True)

df_down = df_down.drop(['대여소명', '대여구분코드', '성별', '운동량', '탄소량'], axis=1)
df_down['대여소번호'] = df_down['대여소번호'].astype('Int64')
```

```python
df_up.to_csv('data/2025년_상반기_공공자전거_이용데이터_1-6.csv', index=False, encoding='utf-8-sig')
df_down.to_csv('data/2025년_하반기_공공자전거_이용데이터_7-12.csv', index=False, encoding='utf-8-sig')

print('데이터 출력 완료!')
```
